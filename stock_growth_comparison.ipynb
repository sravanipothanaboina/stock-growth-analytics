```python
import pandas as pd
import yfinance as yf

# 1. Choose two stocks (Apple and Microsoft)
tickers = ["AAPL", "MSFT"]

# 2. Download historical stock prices for the last year
data = yf.download(tickers, start="2025-01-01", end="2026-01-01")["Close"]

# 3. Show the first 5 rows of data
print("Here is the stock price data:")
display(data.head())
```

```python
import matplotlib.pyplot as plt

# Normalize prices to start at $100 so we can compare growth directly
normalized_data = (data / data.iloc[0]) * 100

# Plot the graph
plt.figure(figsize=(10, 5))
plt.plot(normalized_data['AAPL'], label='Apple (AAPL)', color='blue')
plt.plot(normalized_data['MSFT'], label='Microsoft (MSFT)', color='green')

# Add titles and labels
plt.title('Stock Growth Comparison (Starting at $100)')
plt.xlabel('Date')
plt.ylabel('Normalized Value ($)')
plt.legend()
plt.grid(True)

# Display the chart
plt.show()
```
